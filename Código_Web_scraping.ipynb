```python
import requests
import pandas as pd
```

```python
keyword = "Chile protestas 2019"
api_url='https://api.scrape-it.cloud/scrape/google'
headers = {'x-api-key':'d2a21ab0-02af-40c7-a9d9-6e3ce27ef7e1'}

params ={
    'q': keyword,
    'domain':'google.com',
    'tbm':'nws',
    'tbs':'cdr:1,cd_min:2019,cd_max:2019',
    'sort':'r',
    'gl': 'cl',
    'hl': 'es',
    'num': 10
}

all_news=[]

def fetch(params, label=""):
    r = requests.get(api_url, params=params, headers=headers, timeout=30)
    if r.status_code != 200:
        print(f"Error {label}: HTTP {r.status_code}")
        print("Detalle:", r.text[:500])
        return []
    try:
        data = r.json()
    except ValueError:
        print(f"Error {label}: respuesta no es JSON")
        print("Detalle:", r.text[:500])
        return []
    items = data.get('newsResults', [])
    if not items:
        print(f"Advertencia {label}: 'newsResults' vacío. Claves:", list(data.keys()))
    return items

try:
    # Página 1
    all_news.extend(fetch(params, "solicitud 1"))

    # Página 2 (si el endpoint soporta paginación con 'start')
    params_page2 = params.copy()
    params_page2['start'] = 10
    all_news.extend(fetch(params_page2, "solicitud 2"))

    if all_news:
        df = pd.DataFrame(all_news)
        df.to_csv("Chile_2019_curso.csv", index=False, encoding="utf-8-sig")
        print(f"Listo ✅ {len(df)} filas exportadas a CSV")
    else:
        print("No se obtuvieron noticias.")
except Exception as e:
    print("Error inesperado:", e)

```
