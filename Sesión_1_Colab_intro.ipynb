#Ejercicio en clases

##Pedro tiene 3 manzanas y 6 naranjas y Juan tiene 8 manzanas y 2 naranjas¿ Cuántas frutas tiene cada uno?


```python
import pandas as pd
datos = pd.DataFrame({"Manzanas":[3,8], "Naranjas":[6,2]}, index=["Pedro", "Juan"])
datos
```

```python
datos["total frutas"]=datos['Manzanas']+datos['Naranjas']
print(datos)
```

#Gráfico de barras
